```python
clean_laps = laps[
    (laps['IsAccurate'] == True) &
    (laps['PitInTime'].isna()) &
    (laps['PitOutTime'].isna()) &
    (laps['TrackStatus'] == '1')
].copy()

print(f"Voltas antes: {len(laps)}")
print(f"Voltas depois: {len(clean_laps)}")
```
